```python
"""
Food Content Web Application

This interactive Flask app, designed to run from a Jupyter notebook, lets users explore
nutritional information for a wide variety of foods. The homepage features a searchable
drop-down menu: as you begin typing, matching food items dynamically appear for selection.

Key Features:
    - Homepage with a dynamic drop-down menu of available foods and a header image.
    - Search functionality: select or enter a food to view detailed nutritional information.
    - Friendly error handling: if a food is not found, the app guides the user with informative pages and options to try again or exit.

Usage:
    - Run this notebook in Jupyter Notebook or JupyterLab.
    - See README.md for additional setup, dependencies, or deployment options.
"""
```

```python
import pandas as pd
from flask import Flask, render_template, request
```

```python
# Option 1: If get_food_info is a Python module (.py file) in your project directory, use:
# from get_food_info import load_clean, check_food, try_again, get_input

# Option 2: If get_food_info is a Jupyter notebook (.ipynb), use import-ipynb as follows:
from ipynb.fs.full.get_food_info import load_clean
```

```python
# Load dataset once at startup for efficiency
df=load_clean()
entries=df['Main food description'].to_list()
```

```python
app2 = Flask(__name__)
```

```python
@app2.route('/')
@app2.route('/index')
def index():
    """
    Render the homepage featuring a searchable drop-down menu of available foods and a header image.
    The drop-down allows users to begin typing a food name, displaying matching items in real-time
    for easy selection. The page also includes a decorative header image.
    Returns:
    str: Rendered HTML template for the homepage with food selection capabilities.
    """
    image= 'fruitsvegg.jpg'
    return render_template('index.html', image=image, entries=entries)
    

@app2.route('/food_content', methods=['GET', 'POST'])
def get_content():
    """
    Lookup nutritional content for a requested food.
    If not found, show a friendly error page.
    GET/POST Params:
        food (str): Name of the food to look up.
    Returns:
        Rendered template with food details or 'not found' message.
    """
    food=request.args.get('food')
    if food not in (df['Main food description'].to_list()):
        image='celery3.png'
        return render_template('food_not_found.html', image=image)
    else:
        result=df.loc[df['Main food description']== food,]
        return render_template('food_content.html', food=food, entries=entries, tables=[result.to_html(classes='data', header="true", index=False)])
    
@app2.route('/food_not_found')
def get_content2():
    """
    Handle user responses when a food is not found.
    If user says 'No', show goodbye. If 'Yes', return to homepage.
    Otherwise, show invalid answer page.
    GET Params:
        yes_no (str): User's yes/no response.
    Returns:
        Rendered template based on user response.
    """
    yes_no=request.args.get('yes_no').strip().lower()
    if yes_no not in ('yes', 'no'):
        image='bye3.png'
        return render_template('invalid_answer.html', image=image)
    elif yes_no == 'no':
        image='bye3.png'
        return render_template('good_bye.html', image=image)
    else:
        image='fruitsvegg.jpg'
        return render_template('index.html', image=image, entries=entries)

if __name__ == "__main__":
    # -- Running Flask app --
    # Option 1: For development (default Flask server, use debug mode if needed)
    # app2.run(debug=True, port=5001)
    
    # Option 2: For production deployment, use waitress
    # from waitress import serve
    # serve(app2, host="0.0.0.0", port=8000)
    
    # Default run for this script:
    app2.run(debug=False)
```
